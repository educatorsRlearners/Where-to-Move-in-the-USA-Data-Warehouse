# Drop duplicate cities and states


```python
keep = ['city', 'state', 'zip', 'final_zip_pred', 'AverageMbps', 'alltransit_performance_score']
one_city = df_all.loc[:, keep].drop_duplicates(subset=['city', 'state'])
one_city.shape
```

```python
one_city.sort_values(by=['alltransit_performance_score','AverageMbps'], ascending=False)[:10]
```

```python
one_city.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler(feature_range=(0,10))

original = ['final_zip_pred', 'AverageMbps']
```

```python
one_city[original] = scaler.fit_transform(one_city[original])
one_city[original] = one_city[original].round(1)
```

```python
print(one_city.shape)
one_city.head()
```

```python
one_city['total'] = (one_city["alltransit_performance_score"] + one_city['AverageMbps']) - one_city['final_zip_pred']
```

```python
one_city.loc[:, 'total'] = one_city.loc[:, 'total'].round(2)
one_city.sort_values(by='total', ascending=False)[:20]
```
