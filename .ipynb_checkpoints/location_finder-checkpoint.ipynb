```python
from bs4 import BeautifulSoup
from tqdm import tqdm

import itertools
import json
import pandas as pd 
import pickle
import re
import requests 
```

```python
url = 'https://locations.traderjoes.com/'
page = requests.get(url)
```

```python
page
```

```python
soup = BeautifulSoup(page.content, 'html.parser')
```

```python
# Section of the website which contains list of states w/ TJs
results = soup.find(id='contentbegin')
```

```python
states = [link.get('href') for link in tqdm(results.find_all('a'))]
```

```python
cities = [ ]

for state in tqdm(states): 
    page = requests.get(state)
    soup = BeautifulSoup(page.content, 'html.parser')
    results = soup.find(id='contentbegin')
    names = [link.get('href') for link in results.find_all('a')]
    cities.append(names)
```

```python
flat_cities = list(itertools.chain(*cities))
```

```python
tucson = 'https://locations.traderjoes.com/az/tucson/'
```

```python
def get_main(url):
    page = requests.get(url)
    soup = BeautifulSoup(page.content, 'html.parser')
    results = soup.find(id='contentbegin')
    links = [link.get('href') for link in results.find_all('a')]
    return links
```

```python
store_links = [get_main(city) for city in tqdm(flat_cities)]
```

```python
all_stores = list(set(list(itertools.chain(*store_links))))
```

```python
def get_main(url):
    '''Get the main section of the page which contains the address information'''
    page = requests.get(url)
    soup = BeautifulSoup(page.content, 'html.parser')
    results = soup.find(id='contentbegin')
    return results


def get_info(url): 
    '''Loads the main section and then scrapes the information of interest'''
    raw = get_main(url)
    address = raw.find('div', class_="addressline").get_text().replace('\t', "").strip().split('\n')
    address = [a.strip() for a in address]
    address = list(filter(None, address))
    address = [a for a in address if "," not in a]
    store_name = raw.find('div', class_="h1title").get_text() 
    return [store_name] + address
```

```python
all_store_info = [get_info(store) + [store] for store in tqdm(all_stores)] 
info = ['store_name','street', 'city', 'state', 'zip', 'landline', 'mobile', 'website']

df = pd.DataFrame(all_store_info, columns=info)    
```

```python
df.to_csv('store_info.csv', index=False)
```
