```python
states = ["AL", "AK", "AZ", "AR", "CA", "CO", "CT", "DE", "DC", "FL", "GA", 
          "HI", "ID", "IL", "IN", "IA", "KS", "KY", "LA", "ME", "MD", 
          "MA", "MI", "MN", "MS", "MO", "MT", "NE", "NV", "NH", "NJ", 
          "NM", "NY", "NC", "ND", "OH", "OK", "OR", "PA", "RI", "SC", 
          "SD", "TN", "TX", "UT", "VT", "VA", "WA", "WV", "WI", "WY"]
```

```python
len(states)
```

```python
len(joined_list)
```

```python
joined_list.sort()
```

```python
state_dict = dict(zip(joined_list, states))
```

```python
for k, v in state_dict.items():
    df = pd.read_csv(k)
    df['state'] = v
    df.to_csv('data/01.transformed/transit/'+str(v)+'.csv', index=False)
```

```python
state_files = os.path.join('data/01.transformed/transit', "*.csv")
joined_states = glob.glob(state_files)
```
